now we got a datset with all features we need. now we will extract data from each column and make it to one column.

```python
import ast
```

ast will be used to extract the data from this complex list of dict, we have in our columns

```python
def convert(text):
    L = []
    for i in ast.literal_eval(text):
        L.append(i['name']) 
    return L 
```

```python
movies['genres'] = movies['genres'].apply(convert)
movies.head()
```

```python
movies['keywords'] = movies['keywords'].apply(convert)
movies.head()
```

now from caste coumn we want, character's real name and not all character's names

```python
def convert3(text):
    L = []
    counter = 0
    for i in ast.literal_eval(text):
        if counter < 3:
            L.append(i['name'])
        counter+=1
    return L 
```

```python
movies['cast'] = movies['cast'].apply(convert)
movies.head()
```

```python
movies['cast'] = movies['cast'].apply(lambda x:x[0:3])
```

```python
movies.head()
```

now in crew column we only want directores name, so i will extract only director name 

```python
def fetch_director(text):
    L = []
    for i in ast.literal_eval(text):
        if i['job'] == 'Director':
            L.append(i['name'])
    return L 
```

```python
movies['crew'] = movies['crew'].apply(fetch_director)
```

```python
movies.sample(5)
```

now as our aim is to make MRS, so i will use vectorisation method to do that. so every word will be convert to a number. now for names like "Adam Sandler" it is one name but when vectorising it will become two different numbers. so i will remove sapce between all names.

```python
def collapse(L):
    L1 = []
    for i in L:
        L1.append(i.replace(" ",""))
    return L1
```

```python
movies['cast'] = movies['cast'].apply(collapse)
movies['crew'] = movies['crew'].apply(collapse)
movies['genres'] = movies['genres'].apply(collapse)
movies['keywords'] = movies['keywords'].apply(collapse)
```

```python
movies.head()
```

as i have to make a one column from all these column, and overview column is no in list so i will convert that column's values into list and then concatenate all columns and make a new column

```python
movies.isnull().sum()
```

```python
movies.dropna(inplace=True)
```

```python
movies['overview'] = movies['overview'].apply(lambda x:x.split())
```

```python
movies['tags'] = movies['overview'] + movies['genres'] + movies['keywords'] + movies['cast'] + movies['crew']
```

```python
movies.shape
```

```python
new = movies.drop(columns=['overview','genres','keywords','cast','crew'])
```

```python
new.head()
```

now we got the perfect data and tag column can be removed from the list

```python
new['tags'] = new['tags'].apply(lambda x: " ".join(x))
new.head()
```

now we known there are some words which can be written in different types, like love,loves, actor, actors. these words can create some dulicacy and complexity. so we can make all these words to only one word. and it can be done by nltk library.

```python
import nltk
```

```python
from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()
```

```python
def stem(text):
    y = []
    for i in text.split():
        y.append(ps.stem(i))
    return " ".join(y)
```

```python
new['tags'] = new['tags'].apply(stem)
```

```python
new.head()
```

now our data is ready, we can vectorize it now

```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(max_features=3000,stop_words='english')
```

```python
vector = cv.fit_transform(new['tags']).toarray()
```

```python
vector.shape
```

```python
vector
```

now we have vector dataset with vector values of 4806 rows and 3000 columns. Each row represents a movie and each column represents a unique word from the tags. The values in the matrix indicate the frequency of each word in the tags for each movie. 


we will recommend the movie according to distance. so as we have 3000 dimension data we cant use any distance calculating formula, so i will use cosine distance where it calculate the angle between the vectors

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity = cosine_similarity(vector)
```

```python
similarity
```

now we have distance of all vectors with respect to all others vector.

```python
def recommend(movie):
    index = new[new['title'] == movie].index[0]
    distances = sorted(list(enumerate(similarity[index])),reverse=True,key = lambda x: x[1])
    for i in distances[1:6]:
        print(new.iloc[i[0]].title)
        
    
```

```python
recommend('Avatar')
```
